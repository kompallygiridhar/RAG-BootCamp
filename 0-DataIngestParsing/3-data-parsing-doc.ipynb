### Word Document Processing

```python
from docx import Document as DocxDocument
import os
```

```python
from langchain_community.document_loaders import Docx2txtLoader,UnstructuredWordDocumentLoader
```

```python
## Method1: Using Docx2txtLoader
print("1️⃣ Using Docx2txtLoader")
try:    
    docx_loader=Docx2txtLoader(r"C:\Users\kompa\Desktop\AI_learning\RAG-BootCamp\data\word_files\proposal.docx")
    docs=docx_loader.load()
    print(f"✅ Loaded {len(docs)} document(s)")
    print(f"Content preview: {docs[0].page_content[:200]}...")
    print(f"Metadata: {docs[0].metadata}")

except Exception as e:
    print(f"Error: {e}")
```

```python
##Method2: Using UnstructuredWordDocumentLoader
print("\n2️⃣ Using UnstructuredWordDocumentLoader")

try:
    unstructured_loader=UnstructuredWordDocumentLoader(r"C:\Users\kompa\Desktop\AI_learning\RAG-BootCamp\data\word_files\proposal.docx",mode="elements") ##mode --> preserve the structure of the document
    docs=unstructured_loader.load()

    print(f"✅ Loaded {len(docs)} document(s)")
    for i, doc in enumerate(docs):
        print(f"\nDocument {i+1} Content preview: {doc.page_content[:200]}...")
        print(f"Metadata: {doc.metadata}")
        print(f"Element Type: {doc.metadata.get('category', 'N/A')}")


except Exception as e:
    print(f"Error: {e}")
```
